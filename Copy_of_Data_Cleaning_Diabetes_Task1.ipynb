```python
import pandas as pd

df = pd.read_csv("diabetes (1).csv")

df.head()
```

```python
df.shape
df.columns
df.info()
df.isnull().sum()
```

```python
zero_counts = (df == 0).sum()
zero_counts
```

```python
import numpy as np

columns_with_missing = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

df[columns_with_missing] = df[columns_with_missing].replace(0, np.nan)
```

```python
df.isnull().sum()
```

```python
columns_to_fill = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

for column in columns_to_fill:
    df[column] = df[column].fillna(df[column].median())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df.columns
```

```python
df.dtypes
df.head()
```

```python
df.dtypes
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nDataset shape:")
print(df.shape)
```

```python
df.to_csv("diabetes_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
